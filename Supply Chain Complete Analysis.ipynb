```python
import pandas as pd

# Function to format large monetary values
def format_func(value):
    if value >= 1e6:
        return f'{value / 1e6:.1f}M $'
    elif value >= 1e3:
        return f'{value / 1e3:.1f}K $'
    else:
        return f'{value:.0f} $'


# Create delayed orders dataframe
delayed_df = df[df['Delay'] > 0]

# Calculate business KPIs
metrics = {}

metrics['Total Orders'] = len(df)
metrics['Late Deliveries'] = len(delayed_df)

# Calculate 90th percentile delay
metrics['90% Delay (days)'] = (
    delayed_df['Delay'].quantile(0.90)
    if not delayed_df.empty else 0
)

# Calculate delivery percentages
metrics['On-time Delivery %'] = (
    (len(df) - len(delayed_df)) / len(df) * 100
    if len(df) > 0 else 0
)

metrics['Late Delivery %'] = (
    len(delayed_df) / len(df) * 100
    if len(df) > 0 else 0
)

# Calculate total positive profit
metrics['Total Profit'] = format_func(
    df.loc[df['Order Profit Per Order'] > 0,
           'Order Profit Per Order'].sum()
)

# Calculate negative profit from delayed orders
metrics['Total Loss Due to Delays'] = format_func(
    delayed_df.loc[
        delayed_df['Order Profit Per Order'] < 0,
        'Order Profit Per Order'
    ].sum() * -1
)

# Display business KPIs
print('\n------ Business KPIs ------\n')

for k, v in metrics.items():
    if isinstance(v, (float, int)):
        print(f'{k}: {v:.2f}')
    else:
        print(f'{k}: {v}')
```

```python
#Bottleneck Detection

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Function to calculate delay percentage by category
def compute_delay_pct_by_category(category):

    cat_df = df.groupby(category).agg(
        total_orders=('Delay', 'count'),
        late_orders=('Is_Delayed', 'sum')
    ).reset_index()

    cat_df['delay_pct'] = (
        cat_df['late_orders'] / cat_df['total_orders'] * 100
    )

    cat_df = cat_df.sort_values(
        'delay_pct', ascending=False
    ).head(10)

    return cat_df


# Define categories
categories = [
    'Order Region',
    'Customer Segment',
    'Shipping Mode',
    'Order Status',
    'Type',
    'Department Name'
]


# Create subplot grid
fig, axes = plt.subplots(
    2, 3,
    figsize=(18, 10),
    constrained_layout=True
)

axes = axes.flatten()


# Generate bar plots
for ax, category in zip(axes, categories):

    cat_df = compute_delay_pct_by_category(category)

    sns.barplot(
        data=cat_df,
        x='delay_pct',
        y=category,
        hue=category,
        palette='viridis',
        legend=False,
        ax=ax
    )

    ax.set_title(f'Delay % by {category}')
    ax.set_xlabel('Late Delivery Percentage (%)')
    ax.set_ylabel(category)

    # Add percentage labels
    for i, row in cat_df.reset_index(drop=True).iterrows():

        ax.text(
            row['delay_pct'] + 0.5,
            i,
            f"{row['delay_pct']:.1f}%",
            va='center',
            fontsize=9
        )


plt.suptitle(
    'Late Delivery Percentage by Different Categories',
    fontsize=16,
    fontweight='bold'
)

plt.show()
```

```python
# Top drivers of late delivery by region

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Function to identify top drivers of late delivery by region

def top_drivers_for_region(region):

    # Filter data for selected region
    df_region = df[df['Order Region'] == region].copy()

    if df_region.empty:
        print(f"No data found for region: {region}")
        return

    # Factors to analyze
    drivers = [
        'Shipping Mode',
        'Customer Segment',
        'Department Name',
        'Type',
        'Order Status'
    ]

    all_factors = []

    # Calculate delivery metrics for each factor
    for factor in drivers:

        temp = (
            df_region.groupby(factor)
            .agg(
                total_orders=('Delay', 'count'),
                late_orders=('Is_Delayed', 'sum'),
                avg_delay=('Delay', 'mean')
            )
            .reset_index()
        )

        # Calculate late delivery percentage
        temp['delay_pct'] = (
            temp['late_orders'] / temp['total_orders'] * 100
        )

        # Add driver information
        temp['Driver'] = factor

        temp['Factor_Level'] = (
            factor + " : " + temp[factor].astype(str)
        )

        all_factors.append(
            temp[
                [
                    'Driver',
                    'Factor_Level',
                    'delay_pct',
                    'avg_delay',
                    'total_orders'
                ]
            ]
        )

    # Combine all drivers
    final_df = pd.concat(all_factors, ignore_index=True)

    # Select top 10 drivers
    top_factors = (
        final_df.sort_values(
            'delay_pct',
            ascending=False
        ).head(10)
    )

    # Display top 10 driver metrics
    print(f"\nTop 10 Drivers of Late Delivery in {region}:")
    display(top_factors.round(2))

    # Create horizontal bar chart
    plt.figure(figsize=(12, 7))

    bars = plt.barh(
        top_factors['Factor_Level'],
        top_factors['delay_pct'],
        color='steelblue'
    )

    plt.xlabel("Delay Percentage (%)")
    plt.ylabel("Driver Factors")

    plt.title(
        f"Top Drivers of Late Delivery in {region}",
        fontsize=14,
        fontweight='bold'
    )

    plt.grid(
        True,
        linestyle=':',
        alpha=0.5,
        axis='x'
    )

    plt.gca().invert_yaxis()

    # Add percentage labels
    for bar in bars:

        width = bar.get_width()

        plt.text(
            width + 0.5,
            bar.get_y() + bar.get_height() / 2,
            f'{width:.1f}%',
            va='center',
            fontsize=10
        )

    plt.tight_layout()
    plt.show()


# Execute function for Central Africa
top_drivers_for_region('Central Africa')
```
